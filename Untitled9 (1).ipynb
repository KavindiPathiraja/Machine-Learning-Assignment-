```python
import pandas as pd
import numpy as np

from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from sklearn.model_selection import GridSearchCV
import matplotlib.pyplot as plt
```

```python
import pandas as pd

base_path = "E:/a_4th year_sem2/ML/Assignment/UCI HAR Dataset/UCI HAR Dataset/"

features = pd.read_csv(base_path + "features.txt", sep=r"\s+", header=None)
feature_names = features.iloc[:, 1].tolist()

```

```python
# -----------------------------
# 2. Load training and test data
# -----------------------------
X_train = pd.read_csv(base_path + "train/X_train.txt", sep=r"\s+", header=None)
y_train = pd.read_csv(base_path + "train/y_train.txt", sep=r"\s+", header=None).iloc[:, 0]

X_test = pd.read_csv(base_path + "test/X_test.txt", sep=r"\s+", header=None)
y_test = pd.read_csv(base_path + "test/y_test.txt", sep=r"\s+", header=None).iloc[:, 0]

X_train.columns = feature_names
X_test.columns = feature_names
```

```python
# -----------------------------
# 3. Activity label mapping
# -----------------------------
activity_map = {
    1: "WALKING",
    2: "WALKING_UPSTAIRS",
    3: "WALKING_DOWNSTAIRS",
    4: "SITTING",
    5: "STANDING",
    6: "LAYING"
}
```

```python
# -----------------------------
# 4. Train a baseline Decision Tree
# -----------------------------
dt = DecisionTreeClassifier(
    random_state=42,
    criterion="gini",
    max_depth=10,
    min_samples_split=10,
    min_samples_leaf=5
)

dt.fit(X_train, y_train)
```

```python
# -----------------------------
# 5. Predict and evaluate
# -----------------------------
y_pred = dt.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)
print("Baseline Decision Tree Accuracy:", accuracy)

print("\nClassification Report:")
print(classification_report(
    y_test,
    y_pred,
    target_names=[activity_map[i] for i in sorted(activity_map.keys())]
))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```

```python
# -----------------------------
# 6. Hyperparameter tuning
# -----------------------------
param_grid = {
    "criterion": ["gini", "entropy"],
    "max_depth": [5, 10, 15, 20, None],
    "min_samples_split": [2, 5, 10, 20],
    "min_samples_leaf": [1, 2, 5, 10]
}

grid = GridSearchCV(
    DecisionTreeClassifier(random_state=42),
    param_grid=param_grid,
    cv=5,
    scoring="accuracy",
    n_jobs=-1
)

grid.fit(X_train, y_train)

print("\nBest Parameters:", grid.best_params_)
print("Best Cross-Validation Score:", grid.best_score_)

best_dt = grid.best_estimator_
```

```python
# -----------------------------
# 7. Evaluate tuned model
# -----------------------------
y_pred_best = best_dt.predict(X_test)

best_accuracy = accuracy_score(y_test, y_pred_best)
print("\nTuned Decision Tree Accuracy:", best_accuracy)

print("\nTuned Model Classification Report:")
print(classification_report(
    y_test,
    y_pred_best,
    target_names=[activity_map[i] for i in sorted(activity_map.keys())]
))

print("\nTuned Model Confusion Matrix:")
print(confusion_matrix(y_test, y_pred_best))
```

```python
# -----------------------------
# 8. Top important features
# -----------------------------
importances = pd.DataFrame({
    "feature": X_train.columns,
    "importance": best_dt.feature_importances_
}).sort_values(by="importance", ascending=False)

print("\nTop 15 Important Features:")
print(importances.head(15))
```

```python
# -----------------------------
# 9. Optional: Plot a shallow tree
# -----------------------------
plt.figure(figsize=(20, 10))
plot_tree(
    best_dt,
    max_depth=3,
    filled=True,
    feature_names=X_train.columns,
    class_names=[activity_map[i] for i in sorted(activity_map.keys())],
    fontsize=8
)
plt.title("Decision Tree (First 3 Levels)")
plt.show()
```

```python
path = dt.cost_complexity_pruning_path(X_train, y_train)
ccp_alphas = path.ccp_alphas

dt_list = []
for ccp_alpha in ccp_alphas:
    dt_temp = DecisionTreeClassifier(ccp_alpha=ccp_alpha)
    dt_temp.fit(X_train, y_train)
    dt_list.append(dt_temp)

train_scores = [d.score(X_train, y_train) for d in dt_list]
test_scores = [d.score(X_test, y_test) for d in dt_list]

plt.plot(ccp_alphas, test_scores)
plt.xlabel("Alpha")
plt.ylabel("Accuracy")
plt.title("Pruning Effect")
plt.show()
```
